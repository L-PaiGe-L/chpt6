# 第6章: Matplotlib 坐标轴定制练习题
## 目标
熟练掌握 Matplotlib 坐标轴定制核心技巧，包括刻度控制、图内轴添加、轴脊调整，提升图表专业性。

## 练习 1: 定制刻度位置、格式与样式
### 任务
用给定的 48 小时股票价格数据，绘制折线图(样式 'g-o')，重点实现:
- X 轴主刻度每 12 小时、格式为 YYYY-MM-DD HH:MM
- X 轴次刻度每 2 小时
- 主刻度朝外(蓝色、长10宽2、标签旋转15°)，次刻度朝内(红色、长5)
- 添加网格线，设置合理标题和轴标签

```python
import matplotlib.pyplot as plt
import numpy as np
from datetime import datetime, timedelta

# 准备数据
base = datetime(2025, 1, 1, 0)
x_time = np.array([base + timedelta(hours=i*2) for i in range(25)])
y_price = np.array([100.0, 100.2, 101.0, 100.5, 100.8, 102.0, 101.5, 101.8,
                   101.3, 100.9, 100.0, 99.5, 100.0, 99.8, 100.5, 101.2, 101.0, 101.5, 102.5,
                   103.0, 102.8, 102.5, 102.0, 101.5, 101.8])

# 创建画布
plt.figure(figsize=(12, 6))

# 绘制折线图(样式 'g-o')
plt.plot(x_time, y_price, 'g-o', linewidth=1.5, markersize=4)

# 定制X轴刻度
# 主刻度每12小时
from matplotlib.dates import HourLocator, DateFormatter
ax = plt.gca()
ax.xaxis.set_major_locator(HourLocator(interval=12))
ax.xaxis.set_major_formatter(DateFormatter('%Y-%m-%d %H:%M'))

# 次刻度每2小时
ax.xaxis.set_minor_locator(HourLocator(interval=2))

# 主刻度样式(蓝色、长10宽2、标签旋转15°)
ax.tick_params(axis='x', which='major', length=10, width=2, color='blue', 
               labelrotation=15, labelsize=10)

# 次刻度样式(红色、长5、朝内)
ax.tick_params(axis='x', which='minor', length=5, color='red', direction='in')

# 添加网格线
plt.grid(True, which='major', alpha=0.7)
plt.grid(True, which='minor', alpha=0.3)

# 设置标题和轴标签
plt.title('48小时股票价格走势', fontsize=14, fontweight='bold')
plt.xlabel('时间', fontsize=12)
plt.ylabel('股票价格（元）', fontsize=12)

# 自动调整布局
plt.tight_layout()
plt.show()
```

## 练习 2: 添加图内坐标轴
### 任务
用"大图+小图"展示销量数据:
- 大图(主坐标轴)展示全国总销量(样式 'b-s')
- 小图(嵌入右上角)展示华东区销量(样式 'r-o')
- 主坐标轴位置 [0.1, 0.1, 0.8, 0.8]，小图位置自己调节
- 小图刻度标签缩小，避免拥挤(可取 labelsize=8)
- 设置标题、轴标签和图例

```python
import matplotlib.pyplot as plt
import numpy as np

# 准备数据
x_all = np.arange(1, 13)
y_all_sales = np.array([50, 55, 60, 65, 75, 80, 85, 90, 80, 70, 65, 70])
y_east_sales = np.array([20, 22, 25, 28, 30, 35, 38, 40, 35, 30, 28, 30])

# 创建画布
plt.figure(figsize=(10, 8))

# 主坐标轴 [0.1, 0.1, 0.8, 0.8]
ax_main = plt.axes([0.1, 0.1, 0.8, 0.8])

# 绘制主图(全国总销量，样式 'b-s')
ax_main.plot(x_all, y_all_sales, 'b-s', linewidth=1.5, markersize=6, label='全国总销量')

# 设置主图
ax_main.set_title('全国及华东区销量月度走势', fontsize=14, fontweight='bold')
ax_main.set_xlabel('月份', fontsize=12)
ax_main.set_ylabel('销量（单位：万件）', fontsize=12)
ax_main.set_xlim(0.5, 12.5)
ax_main.set_ylim(40, 95)
ax_main.legend(loc='lower right')
ax_main.grid(True, alpha=0.7)

# 小坐标轴(嵌入右上角)
ax_sub = plt.axes([0.65, 0.65, 0.22, 0.22])

# 绘制小图(华东区销量，样式 'r-o')
ax_sub.plot(x_all, y_east_sales, 'r-o', linewidth=1.2, markersize=4, label='华东区销量')

# 设置小图(刻度标签缩小)
ax_sub.set_xlabel('月份', fontsize=8)
ax_sub.set_ylabel('销量', fontsize=8)
ax_sub.set_xlim(0.5, 12.5)
ax_sub.set_ylim(18, 42)
ax_sub.legend(fontsize=8)
ax_sub.grid(True, alpha=0.5)

# 小图刻度标签缩小
ax_sub.tick_params(axis='both', which='major', labelsize=8)

plt.show()
```

## 练习 3: 隐藏部分轴脊
### 任务
用水平条形图展示编程语言投票数据，重点实现:
- 隐藏顶部和右侧轴脊
- 隐藏 Y 轴刻度线
- (选做)为条形图添加得票数值标签
- 设置标题和轴标签，优化图表简洁度

```python
import matplotlib.pyplot as plt
import numpy as np

# 准备数据
languages = ['Python', 'JavaScript', 'Java', 'C#', 'Go', 'TypeScript']
y_pos = np.arange(len(languages))
votes = np.array([450, 380, 290, 180, 120, 90])

# 创建画布
plt.figure(figsize=(10, 6))

# 绘制水平条形图
bars = plt.barh(y_pos, votes, color='lightblue', edgecolor='black', height=0.6)

# 隐藏顶部和右侧轴脊
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# 隐藏 Y 轴刻度线
ax.tick_params(axis='y', which='both', left=False)

# 设置 Y 轴标签
plt.yticks(y_pos, languages)

# (选做)为条形图添加得票数值标签
for i, v in enumerate(votes):
    plt.text(v + 5, i, str(v), va='center', fontsize=10)

# 设置标题和轴标签
plt.title('编程语言受欢迎程度投票结果', fontsize=14, fontweight='bold')
plt.xlabel('得票数', fontsize=12)
plt.ylabel('编程语言', fontsize=12)

# 调整X轴范围，为数值标签留出空间
plt.xlim(0, max(votes) * 1.1)

# 添加网格线
plt.grid(axis='x', alpha=0.3)

plt.tight_layout()
plt.show()
```

## 练习 4: 移动轴脊到数据中心
### 任务
绘制二次函数 $y=x^{2}$，重点实现:
- 隐藏顶部和右侧轴脊
- 将左右、底部轴脊移动到坐标原点 (0,0)
- 设置合理坐标轴范围，添加图例和标题

```python
import matplotlib.pyplot as plt
import numpy as np

# 准备数据
x = np.linspace(-5, 5, 100)
y = x**2

# 创建画布
plt.figure(figsize=(8, 6))

# 绘制函数图像
plt.plot(x, y, 'r-', linewidth=2, label='$y = x^2$')

# 隐藏顶部和右侧轴脊
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# 将左右、底部轴脊移动到坐标原点 (0,0)
ax.spines['left'].set_position('zero')
ax.spines['bottom'].set_position('zero')

# 隐藏左侧和底部轴脊的默认位置
ax.spines['left'].set_visible(True)
ax.spines['bottom'].set_visible(True)

# 设置坐标轴范围
plt.xlim(-5.5, 5.5)
plt.ylim(-1, 26)

# 设置X轴和Y轴刻度
plt.xticks(np.arange(-5, 6, 1))
plt.yticks(np.arange(0, 26, 5))

# 添加图例和标题
plt.legend(fontsize=12)
plt.title('二次函数 $y = x^2$ 图像', fontsize=14, fontweight='bold')

# 添加网格
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 练习 5: 综合定制(轴脊+刻度)
### 任务
绘制正切函数 $y=\tan(x)$，综合实现:
- 隐藏顶部和右侧轴脊，移动轴脊到原点
- Y 轴范围限制为 [-4, 4](避开无穷大奇点)
- X 轴刻度设为 [-np.pi, -np.pi/2, 0, np.pi/2, np.pi]，标签用 LaTeX 格式(如 $-\pi$)
- 添加网格线、图例和标题

```python
import matplotlib.pyplot as plt
import numpy as np

# 准备数据(避开无穷大奇点：±pi/2附近)
x = np.linspace(-np.pi + 0.1, np.pi - 0.1, 200)
y = np.tan(x)

# 创建画布
plt.figure(figsize=(10, 6))

# 绘制正切函数
plt.plot(x, y, 'b-', linewidth=2, label='$y = \tan(x)$')

# 隐藏顶部和右侧轴脊
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# 移动轴脊到原点
ax.spines['left'].set_position('zero')
ax.spines['bottom'].set_position('zero')

# Y 轴范围限制为 [-4, 4]
plt.ylim(-4, 4)
plt.xlim(-np.pi-0.2, np.pi+0.2)

# X 轴刻度设置(LaTeX格式)
x_ticks = [-np.pi, -np.pi/2, 0, np.pi/2, np.pi]
x_labels = ['$-\pi$', '$-\pi/2$', '$0$', '$\pi/2$', '$\pi$']
plt.xticks(x_ticks, x_labels)

# Y 轴刻度设置
plt.yticks(np.arange(-4, 5, 1))

# 添加网格线
plt.grid(True, alpha=0.3)

# 添加图例和标题
plt.legend(fontsize=12)
plt.title('正切函数 $y = \tan(x)$ 图像', fontsize=14, fontweight='bold')

# 设置LaTeX渲染
plt.rcParams['text.usetex'] = False  # 使用matplotlib的数学文本渲染
plt.rcParams['mathtext.fontset'] = 'stix'

plt.tight_layout()
plt.show()
```
